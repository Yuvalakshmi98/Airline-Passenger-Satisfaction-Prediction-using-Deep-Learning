### **4.2 Outlier and Skewness Identification**
---


````python
# Select numerical columns(Continuos numerical values) for outlier detection
num_cols = ["Age", "Flight Distance", "Departure Delay", "Arrival Delay"]

# Calculate outliers using IQR
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    
    print(f"{col}: {outliers} outliers")
````

---

**SUMMARY:**

* No outliers were detected in Age.
* Flight Distance had 2,855 potential outliers.
* Departure Delay had 18,098 potential outliers.
* Arrival Delay had 17,492 potential outliers.
* Since these values represent valid real-world airline conditions, no rows were removed during outlier treatment.
---

````python
# Skewness and Kurtosis Analysis
# Select numerical columns
numeric_cols = ['Age', 'Flight Distance', 'Departure Delay', 'Arrival Delay']
# Calculate skewness and kurtosis
skewness = df[numeric_cols].skew()
kurtosis = df[numeric_cols].kurtosis()
# Create summary table
skew_kurtosis = pd.DataFrame({
    'Skewness': skewness,
    'Kurtosis': kurtosis
})
skew_kurtosis
````

---
**SUMMARY:**

* **Age** is approximately symmetric, while **Flight Distance** shows moderate positive skewness.
* **Departure Delay** and **Arrival Delay** are highly right-skewed with heavy tails, indicating extreme delay values.
* **No data was removed** based on skewness or kurtosis analysis; valid observations were retained.

### **4.7 Handling Missing Values only for training data**
---

````python

# Calculate median using training data only
arrival_median = X_train["Arrival Delay"].median()

# Apply the same training median to all datasets
for dataset in [X_train, X_val, X_test]:
    dataset["Arrival Delay"] = (
        dataset["Arrival Delay"].fillna(arrival_median)
    )

# Verify missing values
print("Training median:", arrival_median)
print("Training missing values:",
      X_train["Arrival Delay"].isna().sum())
print("Validation missing values:",
      X_val["Arrival Delay"].isna().sum())
print("Test missing values:",
      X_test["Arrival Delay"].isna().sum())
````

---
**Method Used:** Median Imputation

**Justification:**

* Arrival Delay is a numerical variable with only 393 missing values. 
* Median imputation was used because flight delay data may contain extreme values, and the median is less affected by outliers than the mean.

#### **8.1 Build - Improved ANN model**

````python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

# Set random seed
tf.keras.utils.set_random_seed(42)

# Build improved ANN
improved_model = Sequential([
    Dense(128, activation='relu', input_shape=(X_train.shape[1],)),
    BatchNormalization(),
    Dropout(0.3),

    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.2),

    Dense(32, activation='relu'),

    Dense(1, activation='sigmoid')
])

# Compile the model
improved_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Display model architecture
improved_model.summary()
````

----
**SUMMARY**

- **Improved the baseline ANN** by incorporating **Batch Normalization** after the first two hidden layers to stabilize training and improve convergence.
- **Retained the 128, 64, and 32 neuron architecture** with ReLU activation and dropout rates of 0.3 and 0.2 to reduce overfitting.
- **Explicitly configured the Adam optimizer** with a learning rate of 0.001, along with binary cross-entropy loss for binary classification.
- **Maintained reproducibility** by setting the random seed to 42.
- Overall, the improved ANN introduces **Batch Normalization** while retaining the baseline architecture and regularization strategy.

---
#### **8.2 Train the improved ANN**
---

````python
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# Stop training when validation loss stops improving
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

# Reduce learning rate when validation loss stops improving
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=3,
    min_lr=0.00001
)

# Train the model
improved_history = improved_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=64,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)
````

----
**SUMMARY**

- **Early Stopping:** Stops training when validation loss does not improve for 5 epochs and restores the best model weights.
- **ReduceLROnPlateau:** Reduces the learning rate by 50% when validation loss does not improve for 3 epochs.
- **Benefits:** Helps prevent overfitting, optimize training time, and potentially improve model generalization.

### **8.4. ANN Model-Hyperparameter Tuning**
----

#### **Testing different hyperparameters**

````python
import tensorflow as tf
import pandas as pd

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Hyperparameter configurations
configurations = [
    {"neurons": 64, "dropout": 0.2, "learning_rate": 0.001},
    {"neurons": 128, "dropout": 0.3, "learning_rate": 0.001},
    {"neurons": 256, "dropout": 0.4, "learning_rate": 0.0005}
]

tuning_results = []

for config in configurations:

    print("\nTesting configuration:", config)

    # Build model
    model = Sequential([
        Input(shape=(X_train.shape[1],)),

        Dense(config["neurons"], activation="relu"),
        BatchNormalization(),
        Dropout(config["dropout"]),

        Dense(32, activation="relu"),
        Dense(1, activation="sigmoid")
    ])

    # Compile model
    model.compile(
        optimizer=Adam(
            learning_rate=config["learning_rate"]
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    # Early stopping
    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    )

    # Train model
    tuned_history = model.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        epochs=30,
        batch_size=64,
        callbacks=[early_stopping],
        verbose=0
    )

    # Validation predictions
    y_val_prob = model.predict(X_val, verbose=0)
    y_val_pred = (y_val_prob >= 0.5).astype(int).ravel()

    # Calculate validation metrics
    tuning_results.append({
        "Neurons": config["neurons"],
        "Dropout": config["dropout"],
        "Learning Rate": config["learning_rate"],
        "Accuracy": accuracy_score(y_val, y_val_pred),
        "Precision": precision_score(y_val, y_val_pred),
        "Recall": recall_score(y_val, y_val_pred),
        "F1 Score": f1_score(y_val, y_val_pred)
    })

# Display comparison
tuning_df = pd.DataFrame(tuning_results)
tuning_df = tuning_df.sort_values(
    by="F1 Score", ascending=False
).reset_index(drop=True)

tuning_df
````

---

**SUMMARY: ANN Hyperparameter Tuning**

- **Best Configuration:** The configuration with **256 neurons, 0.4 dropout, and a 0.0005 learning rate** achieved the highest accuracy (**95.05%**) and F1-score (**0.9421**) among the three tested configurations.

- **Performance Comparison:** Accuracy ranged from **94.94% to 95.05%**, while F1-scores ranged from **0.9409 to 0.9421**, indicating relatively stable performance across configurations.

- **Precision vs. Recall:** Precision (**95.29%–95.85%**) was higher than recall (**92.62%–92.97%**) across all configurations, indicating better precision than positive-class coverage.

- **Model Capacity:** Increasing the number of neurons from 64 to 256 did not produce a substantial performance improvement, suggesting that a larger network may not be necessary.

- **Efficiency:** The **64-neuron configuration** offers a smaller architecture with performance close to that of the best configuration.

**Conclusion:**

Among the three tested configurations, **256 neurons, 0.4 dropout, and a 0.0005 learning rate** performed best, achieving **95.05% accuracy and a 0.9421 F1-score**. However, the performance differences are small, so further comparison using validation results and training efficiency can help determine the most practical configuration.

### **8.6. Optimization Techniques**

````python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

tf.keras.utils.set_random_seed(42)

# Keep your tuned hyperparameters
neurons = 256
dropout_rate = 0.4
learning_rate = 0.0005

# Build fine-tuned ANN model
finalann_model = Sequential([
    Input(shape=(X_train.shape[1],)),

    Dense(neurons, activation='relu'),
    BatchNormalization(),
    Dropout(dropout_rate),

    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(dropout_rate),

    Dense(1, activation='sigmoid')
])

# Compile model
finalann_model.compile(
    optimizer=Adam(learning_rate=learning_rate),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Early stopping
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=8,
    restore_best_weights=True,
    verbose=1
)

# Reduce learning rate when validation loss plateaus
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=3,
    min_lr=0.00001,
    verbose=1
)
````

- **Applied 40% Dropout:** Randomly disables neurons during training to reduce overfitting and improve generalization.

- **Added Early Stopping:** Monitors validation loss and restores the best model weights when validation performance stops improving.
- **Added ReduceLROnPlateau:** Reduces the learning rate by half when validation loss stops improving, allowing finer adjustments during training. The batch size was also reduced to 32 to update model weights more frequently.
